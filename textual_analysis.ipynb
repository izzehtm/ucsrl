```python
from fkscore import fkscore as f
from pathlib import Path
import pandas as pd
import re
```

```python
# configure inputs
input_dirs = [
    Path(r".\Data\text_data_cleaned\bodies"),
    Path(r".\Data\text_data_cleaned\unions"),
]
```

```python
# create dataframes
dfs = {}

for parent in input_dirs:
    for folder in parent.iterdir():
        if not folder.is_dir():
            continue

        # initialise data array
        data = []

        for file in folder.glob("caption_*.txt"):
            caption = int(re.search(r"caption_(\d+)\.txt", file.name).group(1))

            with open(file, "r", encoding="utf-8") as f:
                text = f.read()

            # get fkscore
            score = fkscore(text)

            # write caption to row
            row = {"caption": caption}

            # write score to row
            row.update(score.score)

            # append row to data array
            data.append(row)

        # convert data array to dataframe
        dfs[folder.name] = (
            pd.DataFrame(data)
              .sort_values("caption")
              .reset_index(drop=True)
        )
```
