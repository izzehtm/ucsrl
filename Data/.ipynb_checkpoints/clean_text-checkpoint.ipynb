```python
import json
from pathlib import Path
```

```python
# configure directories
# inputs
bodies_dir = Path("./text_data_raw/bodies")
unions_dir = Path("./text_data_raw/unions")

# output
output_base = Path("./text_data_cleaned")
```

```python
def extract_captions(input_dir: Path, output_dir: Path):
    """Parses input JSON for caption data"""
    
    # make output directory if required
    output_dir.mkdir(parents=True, exist_ok=True)

    for json_file in input_dir.glob("*.json"):
        # make output directory
        account_name = json_file.stem
        account_output = output_dir / account_name
        account_output.mkdir(parents=True, exist_ok=True)

        with open(json_file, "r", encoding="utf-8") as f:
            data = json.load(f)

        # get posts
        posts = data.get("data", {}).get("posts", [])

        # count captions for a) naming files, b) verification message
        caption_num = 1
        for post in posts:
            # get "caption" values from data
            caption = post.get("caption")

            # handle for missing/empty captions (skip)
            if not caption:
                continue

            # create output file
            out_file = account_output / f"caption_{caption_num}.txt"

            # write to output file
            with open(out_file, "w", encoding="utf-8") as f:
                f.write(caption.strip())

            caption_num += 1

        print(f"{json_file.name}: wrote {caption_num - 1} captions") #should probably have just 0-indexed my files!
```

```python
# process both datasets
extract_captions(bodies_dir, output_base / "bodies")
extract_captions(unions_dir, output_base / "unions")
```
