```python
from pathlib import Path
import pandas as pd
from readability_to_grade import r2g
```

```python
# initialise base directory
base_dir = Path(r"Data\analysed_text\unions")
```

```python
# map subfolder to union
folder_to_union = {
    "anmf_nursing_midwifery_union": "ANMF",
    "asunion": "ASU",
    "awunion": "AWU",
    "cfmeunational": "CFMEU",
    "cpsunion": "CPSU",
    "etu_australia": "ETU",
    "fsu_australia": "FSU",
    "professionals.australia": "PA",
    "rtbuaustralia": "RTBU",
    "sdaunion": "SDA",
    "theamwu": "AMWU",
    "twuaus": "TWU",
    "unitedworkersoz": "UWU",
    "withmeaa": "MEAA",
}
```

```python
# initialise results array
rows = []

for csv_file in base_dir.glob("*.csv"):
    # read df
    df = pd.read_csv(csv_file)

    # get mean readability
    avg = df["readability"].mean()

    # append results to results array
    rows.append({
        "union": folder_to_union[csv_file.stem],
        "average_readability_score": avg,
        "average_read_grade": r2g(avg)
    })

# convert results array to dataframe
readability_df = pd.DataFrame(rows)
```

```python
# initialise df of results of education by industry analysis
education = pd.read_csv("Data/education_by_industry_results.csv")

# merge readability of unions social media captions with average education levels of industry
result = education.merge(readability_df, on="union", how="left")

# write merged df to csv
result.to_csv(
    "Data/readability_vs_industry_education.csv",
    index=False
)
```

```python
# get summary statistics (uncomment print statements to see results)
median_corr = result['median'].corr(result['average_readability_score'])
mean_corr = result['mean'].corr(result['average_readability_score'])
#print("Median vs Readability Correlation: " + str(median_corr))
#print("Mean vs Readability Correlation: " + str(mean_corr))

counts = result['average_read_grade'].value_counts()
#print("\nValue Counts:")
#print(counts)
```
