```python
from pathlib import Path
import pandas as pd
from readability_to_grade import r2g
```

```python
# initialise base directory
base_dir = Path("./Data/analysed_text/bodies")
```

```python
# initialise results array
rows = []

for csv_file in base_dir.glob("*.csv"):
    df = pd.read_csv(csv_file)

    # append average readability score by union
    rows.append({
        "union": csv_file.stem,
        "average_readability_score": df["readability"].mean(),
        "n_posts": len(df)
    })

```

```python
# convert results to dataframe
readability_df = pd.DataFrame(rows)

# apply readability to grade mapping
readability_df["average_read_grade"] = (
    readability_df["average_readability_score"]
        .apply(r2g)
)
```

```python
# save results to csv
readability_df.to_csv('./Data/readability_by_body.csv')

#print(readability_df) <--- uncomment to see results
```
